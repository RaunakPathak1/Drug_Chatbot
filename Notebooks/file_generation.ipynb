```python
from dotenv import load_dotenv
import os
from openai import OpenAI
from IPython.display import Markdown, display
from pathlib import Path
import time
import random
```

```python
load_dotenv(override=True)
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')

if openrouter_api_key:
    print(f"OpenRouter API Key exists and begins {openrouter_api_key[:3]}")
else:
    print("OpenRouter API Key not set (and this is optional)")

openrouter_url = "https://openrouter.ai/api/v1"

openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
```

```python
MODEL = 'x-ai/grok-4.1-fast'
file_dir_paracetamol = r'C:\Users\rauna\projects\My Projects\Drug Chatbot\Synthetic_files\Paracetamol'
file_dir_insulin = r'C:\Users\rauna\projects\My Projects\Drug Chatbot\Synthetic_files\Insulin'
```

```python
topics = [
    'Full name and identity details',
'Demographics such as age, gender, location, nationality',
'Educational background including degrees, institutions, certifications',
'Current professional title and role',
'Career history and previous positions',
'Technical, functional, and soft skills',
'Domain expertise and areas of specialization',
'Key achievements and notable projects',
'Work style and personality traits',
'Communication style and collaboration habits',
'Tools, software, and technologies regularly used',
'Industry knowledge and familiarity with standards or regulations',
'Short-term and long-term professional goals',
'Personal motivations and core values',
'Challenges, weaknesses, or growth areas',
'Publications, patents, or conference contributions',
'Professional network such as mentors, stakeholders, or teams',
'Hobbies, personal interests, and lifestyle context',
'Work-life balance preferences',
'Digital footprint such as portfolio links, personal website, or summary bio'
]
```

```python
len(topics)
```

```python
for i in range(25):
        if i < len(topics):
            # Use single topics for first N files
            topic = topics[i]
            prompt = f"Write {topic}. Make it engaging and informative."
            print(prompt)
        else:
            # Combine 2-3 random topics for remaining files
            combined_topics = random.sample(topics, k=min(2, len(topics)))
            topic = " and ".join(combined_topics)
            prompt = f"Write about {topic}. Make it engaging and informative, connecting these themes naturally."
            print(prompt)
        
```

```python
def generate_files_paracetamol(num_files=1, output_dir=file_dir_paracetamol, delay=1.0):
    print(f"Generating {num_files} files\n")

    for i in range(num_files):
        filename = f"Paracetamol_{i+1:02d}.txt"
        filepath = os.path.join(output_dir, filename)
        if i < len(topics):
            topic = topics[i]
            prompt = f"Write about {topic}. Make it engaging and informative."

        else:
            combined_topics = random.sample(topics, k=min(2, len(topics)))
            topic = " and ".join(combined_topics)
            prompt = f"Write about {topic}. Make it engaging and informative, connecting these themes naturally."

        try:
            response = openrouter.chat.completions.create(
                model=MODEL, 
                messages=[
                    {
                        "role": "system",
                        "content": '''you are an expert in creating synthetic data for a person named Paracetamol "
                                    Here are some details about Paracetamol: 
                                    Name : Paracetamol Chad
                                    Age: 30
                                    profession: LLM Researcher
                                    Location: New York
                                    Hobbies: hiking, photography, cooking
                                    Industry: Technology
                                    Use these details to create engaging and informative content based on the given topics.
                        '''
                    },
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                max_tokens=750,
                temperature=0.7
            )
            content = response.choices[0].message.content
            with open(filepath, 'w', encoding='utf-8') as f:
                f.write(f"Topic: {topic}\n\n")
                f.write(content)
            
            print(f"Generated: {filename}")

            # Rate limiting
            if i < num_files - 1:
                time.sleep(delay)
            
        except Exception as e:
            print(f"✗ Error generating {filename}: {str(e)}")
    
    print(f"\n✓ All files saved to '{output_dir}/' directory")
```

```python
generate_files_paracetamol(num_files=25)
```
