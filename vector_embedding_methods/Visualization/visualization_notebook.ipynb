```python
full_doc_collection = chroma_client.get_collection('new_data_full_document_embeddings')
sentence_doc_collection = chroma_client.get_collection('new_chunked_by_sentence_collection')
```

```python
# Full doucment embeddings retrieval

full_result = full_doc_collection.get(include=['embeddings', 'documents', 'metadatas'])
full_vectors = np.array(full_result['embeddings'])
full_documents = full_result['documents']
full_metadatas = full_result['metadatas']
full_doc_types = [metadata['person'] for metadata in full_metadatas]
full_colors = [['blue', 'red'][['Paracetamol', 'Insulin'].index(t)] for t in full_doc_types]
```

```python
print(full_metadatas)
```

```python
# Sentence doucment embeddings retrieval

sentence_result = sentence_doc_collection.get(include=['embeddings', 'documents', 'metadatas'])
sentence_vectors = np.array(sentence_result['embeddings'])
sentence_documents = sentence_result['documents']
sentence_metadatas = sentence_result['metadatas']
sentence_doc_types = [metadata['person'] for metadata in sentence_metadatas]
sentence_colors = [['blue', 'red'][['Paracetamol', 'Insulin'].index(t)] for t in sentence_doc_types]
```

```python
# Full doucment embeddings retrieval visualization
# We humans find it easier to visalize things in 2D!
# Reduce the dimensionality of the vectors to 2D using t-SNE
# (t-distributed stochastic neighbor embedding)

tsne = TSNE(n_components=2, random_state=42)
reduced_vectors = tsne.fit_transform(full_vectors)

# Create the 2D scatter plot
fig = go.Figure(data=[go.Scatter(
    x=reduced_vectors[:, 0],
    y=reduced_vectors[:, 1],
    mode='markers',
    marker=dict(size=5, color=full_colors, opacity=0.8),
    text=[f"Type: {t}<br>Text: {d[:100]}..." for t, d in zip(full_doc_types, full_documents)],
    hoverinfo='text'
)])

fig.update_layout(title='2D Chroma Vector Store Visualization',
    scene=dict(xaxis_title='x',yaxis_title='y'),
    width=800,
    height=600,
    margin=dict(r=20, b=10, l=10, t=40)
)

fig.show()
```

```python
# Sentence doucment embeddings retrieval visualization
# We humans find it easier to visalize things in 2D!
# Reduce the dimensionality of the vectors to 2D using t-SNE
# (t-distributed stochastic neighbor embedding)

tsne = TSNE(n_components=2, random_state=42)
reduced_vectors = tsne.fit_transform(sentence_vectors)

# Create the 2D scatter plot
fig = go.Figure(data=[go.Scatter(
    x=reduced_vectors[:, 0],
    y=reduced_vectors[:, 1],
    mode='markers',
    marker=dict(size=5, color=sentence_colors, opacity=0.8),
    text=[f"Type: {t}<br>Text: {d[:100]}..." for t, d in zip(sentence_doc_types, sentence_documents)],
    hoverinfo='text'
)])

fig.update_layout(title='2D Chroma Vector Store Visualization',
    scene=dict(xaxis_title='x',yaxis_title='y'),
    width=800,
    height=600,
    margin=dict(r=20, b=10, l=10, t=40)
)

fig.show()
```
